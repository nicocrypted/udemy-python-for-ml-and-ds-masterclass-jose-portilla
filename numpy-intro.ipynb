```python
arr = np.arange(0,11)
print(arr)
slice_of_arr = arr[0:5]
print(slice_of_arr)
slice_of_arr[:] = 99
print(slice_of_arr)
print(arr) # Notice how the original array has also changed. This is because slice_of_arr is a view of arr, not a copy.
```

```python
arr = np.arange(0,11)
print(arr)
arr_copy = arr.copy() # Now arr_copy is a copy of arr, not a view.
slice_of_arr = arr_copy[0:5]
print(slice_of_arr)
slice_of_arr[:] = 99
print(slice_of_arr)
print(arr)
```
