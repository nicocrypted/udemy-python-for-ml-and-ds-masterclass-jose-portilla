```python
import numpy as np
import pandas as pd
```

```python
# Apply method: Build a custom function and apply it to a Pandas Series
# Example:
def last_four(num):
    '''Grab the last for digits of the num value.'''
    return str(num)[-4:]
df = 'Some Pandas dataframe'
df['CC Number'].apply(last_four)

# Another example with a more complex function:
def yelp(price):
    if price < 10:
        return '$'
    elif price >= 10 and price < 30:
        return '$$'
    else:
        return '$$$'
df['yelp'] = df['total_bill'].apply(yelp)

#Apply method using lambda function:
df['total_bill'].apply(lambda num: num*2)

#Apply function on multiple columns:
def quality(total_bill, tip):
    if tip/total_bill > 0.25:
        return 'Generous'
    else:
        return 'Other'

df['Quality'] = df[['total_bill', 'tip']].apply(lambda df: quality(df['total_bill'], df['tip']), axis = 1)

# Use vectorize to speed up computations
df['Quality'] = np.vectorize(quality)(df['total_bill'], df['tip'])
```
